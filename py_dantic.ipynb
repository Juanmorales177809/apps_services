```python
from pydantic import BaseModel, ValidationError, Field
```

```python
class Person(BaseModel):
    name : str 
    last_name : str
    age : int
```

```python
p = Person(name="Juan", last_name="Morales", age=30)
```

```python
str(p)
```

```python
repr(p)
```

```python
p.model_fields
```

```python
from pydantic import ValidationError
```

```python
try:
    Person(last_name="Juan")
except ValidationError as ex:
    print(ex)
```

```python
data_json = """ {
    "name": "Isaac",
    "last_name": "Newton",
    "age": 30
    }
    """
```

```python
try:
    k = Person.model_validate(data_json)
except ValidationError as ex:
    print(ex)
```
